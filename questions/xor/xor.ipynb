```python
import numpy as np
from sklearn.neural_network import MLPClassifier

X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0, 1, 1, 0])

print("--- XOR Dataset ---")
print("Inputs (X):\n", X)
print("Expected Outputs (y):\n", y)

model = MLPClassifier(
    hidden_layer_sizes=(2,),
    activation='logistic',
    solver='lbfgs',
    max_iter=2000,
    random_state=42
)

print("\nTraining the network of perceptrons...")
model.fit(X, y)
print("Training complete.")

print("\n--- Testing the Trained Network ---")

predictions = model.predict(X)

print(f"Input [0, 0] -> Expected 0, Model Predicted: {predictions[0]}")
print(f"Input [0, 1] -> Expected 1, Model Predicted: {predictions[1]}")
print(f"Input [1, 0] -> Expected 1, Model Predicted: {predictions[2]}")
print(f"Input [1, 1] -> Expected 0, Model Predicted: {predictions[3]}")

accuracy = model.score(X, y)
print(f"\nModel Accuracy: {accuracy * 100:.2f}%")
```
